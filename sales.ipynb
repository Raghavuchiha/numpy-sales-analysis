```python
import numpy as np
```

## Sales Analysis

## import data


```python
months = np.array([
    "Jan", "Feb", "Mar", "Apr",
    "May", "Jun", "Jul", "Aug"
])

sales = np.array([
    1200, 1500, 1100, 1800,
    2100, 1700, 2500, 2300
])

expenses = np.array([
    800, 900, 850, 1000,
    1200, 950, 1400, 1300
])
```

```python
###calculate total sales

total_sales = np.sum(sales)
print("total sales : ",total_sales)
```

```python
avg_sales = np.average(sales)
print("avg sales : ",avg_sales)
```

```python
highest_sale = np.max(sales)
print("highest sale : ",highest_sale)
```

```python
lowest_sale = np.min(sales)
print("lowest sale : ",lowest_sale)
```

```python
profit = sales - expenses
print("profit : ",profit)
```

```python
total_profit = np.sum(sales - expenses)
print("total profit : ",total_profit )
```

```python
profit = sales - expenses
max_index = np.argmax(profit)
print("Monthly Max Profit:")
print(months[max_index], ":", profit[max_index])
```

```python
top_months = months[sales > 2000]
print("Top Sales Months:", top_months)
```

```python
below_avg_months = months[sales < avg_sales]
print("Below Average Sales Months:", below_avg_months)
```

```python
growth = sales[1:] - sales[:-1]


for i in range(len(growth)):
    print(months[i + 1], ":", growth[i])
```

```python
highest_growth_index = np.argmax(growth)

print("Highest Growth Month:",
      months[highest_growth_index + 1],
      "-",
      growth[highest_growth_index])
```

```python
# Task 12 — Calculate Profit Percentage

profit = sales - expenses

profit_percentage = (profit / sales) * 100

print("Monthly Profit Percentage:\n")

for i in range(len(months)):
    print(months[i], ":", round(profit_percentage[i], 2), "%")
```

```python
# Task 14 — Create Performance Labels Based on Sales

performance_labels = np.where(
    sales > 2500,
    "Excellent",
    np.where(sales >= 1800, "Good", "Average")
)

print("Monthly Performance Labels:\n")

for i in range(len(months)):
    print(months[i], ":", performance_labels[i])
```
